```python
from collections import Counter
import os
import pandas as pd
```

```python
def majority_vote_anomalies(folder_path):
    anomaly_counts = {}
    file_count = 0

    # Iterate through all CSV files
    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_count += 1
            file_path = os.path.join(folder_path, filename)
            df = pd.read_csv(file_path)

            # Ensure column names match expected format
            if "Date" not in df.columns or "Is_Anomaly" not in df.columns:
                print(f"Skipping {filename}: missing required columns.")
                continue
            
            # Get anomaly dates in this file
            anomaly_dates = df[df["Is_Anomaly"] == True]["Date"]

            # Count anomalies
            for date in anomaly_dates:
                anomaly_counts[date] = anomaly_counts.get(date, 0) + 1

    # Determine majority threshold
    threshold = file_count / 4

    # Get dates occurring as anomalies in majority of parks
    majority_dates = [
        date for date, count in anomaly_counts.items()
        if count > threshold
    ]

    return majority_dates
```

```python
folder = "knn_anomalies/results/combined"
majority_anomalies = majority_vote_anomalies(folder)

print("Dates marked as anomalies in a majority of parks:")
for d in majority_anomalies:
    print(d)
```
